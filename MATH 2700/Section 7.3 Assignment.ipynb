### Sums

We can compute sums using the `sum()` command.

To compute the sum $\displaystyle\sum_{k=a}^b f(k)$, where $f(k)$ is some function of $k$, we first tell Sage that we plan to use $k$ as an index with `k=var('k')`.

Then we use the syntax `sum(f(k),k,a,b)`.

For example, we compute the sum $\displaystyle\sum_{k=1}^4 k^2$ below.  (Note that one can enter $k^2$ as either `k^2` or `k**2`.)

Check that the answer is correct!

```python
k=var('k')
sum(k^2,k,1,4)
```
